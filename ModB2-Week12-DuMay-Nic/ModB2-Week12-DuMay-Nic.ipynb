# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt

# Data example from Chapter 8 logic
categories = ['Solar', 'Wind', 'Natural Gas', 'Hydro', 'Coal']
values = [78, 92, 60, 65, 40]
highlight_index = 1  # Emphasize Wind

# Create color scheme: highlight Wind
colors = ['#d3d3d3'] * len(categories)
colors[highlight_index] = '#d62728'  # Red highlight for emphasis

# Plot
plt.figure(figsize=(10, 6))
bars = plt.barh(categories, values, color=colors)
plt.gca().invert_yaxis()  # Highest value on top

# Add annotation
plt.text(92, highlight_index, 'Wind leads in current capacity', va='center', ha='left', fontsize=10, color='#d62728')

# Titles and axes
plt.title('Energy Source Capacity by Type')
plt.xlabel('Capacity Index')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 4. Storytelling With Data Plot (Chapter 8: Pulling It All Together)

This chart is modeled after a horizontal bar chart featured in Chapter 8 of *Storytelling with Data* (pp. 187–205). The goal here was not to show all data equally, but to guide the viewer toward a key insight — that wind energy currently leads in capacity. This was done through strategic use of color, text annotations, and minimal design.

This approach reflects a shift from exploratory data analysis to explanatory communication. Rather than showing all energy sources with equal weight, I emphasized wind using red — a visual preattentive attribute — and included a callout to drive the story forward.

This visual builds on my previous analysis by transforming insights into a clean, actionable format that could be used in an executive briefing or policy proposal. It demonstrates how design choices, when applied intentionally, can support more effective data-driven storytelling.
